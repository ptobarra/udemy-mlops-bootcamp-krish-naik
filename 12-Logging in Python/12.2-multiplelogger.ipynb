#### Logging with multiple Loggers
You can crete multiple loggers for different parts of your application.

```python
import logging
```

```python
## create a logger for module 1
logger1 = logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)
```

```python
## create a logger for module 2
logger2 = logging.getLogger("module2")
logger2.setLevel(logging.WARNING)
```

```python
# Configure logging settings
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    force=True,
)
```

```python
## log message with different loggers
logger1.debug("This is a debug message for module 1.")
logger2.warning("This is a warning message for module 2.")
logger2.error("This is an error message.")
```
