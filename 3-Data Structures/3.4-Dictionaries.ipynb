#### introduction to dictionaries
dictionaries are unordered collections of items. they store data in key-value pairs. keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.

```python
## creating dictionaries
#### list [], tuple ()
empty_dict = {}
print(type(empty_dict))
```

```python
empty_dict = dict()
print(type(empty_dict))
```

```python
student = {"name": "Krish", "age": "32", "grade": 24}
print(student)
```

```python
print(type(student))
```

```python
# Single key is always used
student = {"name": "Krish", "age": "32", "name": 24}
print(student)
```

```python
## accessing dictrionary elements
student = {"name": "Krish", "age": "32", "grade": 'A'}
print(student)
```

```python
## acccessing dictionary elements
print(student['grade'])
print(student['age'])
```

```python
## accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
```

```python
print(student.get('last_name', "Not available"))
```

```python
## modifying dictionary elements
## dictionaries are mutable. you can add, update or delete elements.

print(student)
```

```python
student["age"] = 33 # updated the value for the key
print(student)
student["address"] = "India" # added a new key and value
print(student)
```

```python
del student['grade'] ## delete the key and value pair
print(student)
```

```python
## dictionary methods

keys = student.keys() ## get all the keys
print(keys)
values = student.values() ## get all the values
print(values)
```

```python
items = student.items() ## get all key value pairs - a list of tuples
print(items)
```

```python
## shallow copy
student_copy = student
print(student)
print(student_copy)
```

```python
student["name"] = "Krish1"
print(student)
print(student_copy)
```

```python
student["name"] = "Krish2"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() ## shallow copy
print(student)
print(student_copy1)
```

```python
student["name"] = "Krish3"
print(student)
print(student_copy1)
```

```python
### iterating over dictionaries
## you can use loops to iterate over dictionaries, keys, values, or items

## iterating over keys
for key in student.keys():
    print(key)
```

```python
## iterate over values
for value in student.values():
    print(value)
```

```python
## iterate over key value pairs
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
## nested dictionaries
.students = {
    'student1': {"name": "Krish", "age": 32},
    'student2': {"name": "Peter", "age": 35},
}

print(students)
```

```python
## access nested dictionaries elements
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## iterating over dictionaries

for student_id, student_info in students.items():
    print(f"{student_id}: {student_info}")
    for key, value in student_info.items():
        print(f"{key}: {value}")
```

```python
## dictionary comprehension
squares = {x: x**2 for x in range(5)}
print(squares)
```

```python
## condition dictionary comprehension
evens = {x: x**2 for x in range(10) if x % 2 == 0}
print(evens)
```

```python
## practical examples

## use a dictionary to count the frequency of elements in list

numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for number in numbers:
    if number in frequency:
    # if number in frequency.keys():
        frequency[number] += 1
    else:
        frequency[number] = 1

print(frequency)
```

```python
## merge 2 dictionaries into one

dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}

merged_dict = {**dict1, **dict2}
print(merged_dict)
```
