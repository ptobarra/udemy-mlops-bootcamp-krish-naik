#### standard library overview
python's standard library is a vast collection of modules and packages that come bundled with python, porviding a wide range of functionalities out of the box. here's an overview of some of the most commonly used modules and packages in the python standard library.

```python
import array
arr = array.array('i', [1,2,3,4])
print(arr)
```

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
## random

import random
print((random.randint(1, 10)))
print((random.choice(['apple', 'banana', 'cherry'])))
```

```python
### file and directory access

import os

print(os.getcwd())
```

```python
os.makedirs('test_dir')
```

```python
path = "example.txt"

# Create (or overwrite) a file and write to it
with open(path, "w") as f:
    f.write("Hello, world!\n")

print(os.path.exists(path))  # True
```

```python
## high level operations on files and collection of files
import shutil
shutil.copyfile('source.txt', 'destination.txt')
```

```python
## Data Serialization
import json
data = {'name': 'Krish', 'age': 25}

json_str = json.dumps(data)
print(json_str)
print(type(json_str))
parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## csv

import csv

with open('example.csv', mode = 'w', newline = '') as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(['Krish', 32])

with open('example.csv', mode = 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
## datetime
from datetime import datetime, timedelta

now = datetime.now()
print(now)

yesterday = now - timedelta(days = 1)
print(yesterday)
```

```python
## time
import time
print(time.time())
time.sleep(2)
print(time.time())
```

```python
## regular expression
import re

pattern = r'\d+'
text = 'There are 123 apples 456'
match = re.search(pattern, text)
print(match.group())
```
