#### importing modules in python: modules and packages
in python, modules and packages help organize and reuse code. here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
```

```python
from math import sqrt, pi
print(math.sqrt(16))
print(math.sqrt(25))
print(math.sqrt(pi))
```

```python
import numpy as np
```

```python
import numpy as np
```

```python
np.array(([1,2,3,4]))
```

```python
from math import *
print(sqrt(16))
print(pi)
```

```python
from package.maths import addition

addition(2, 3)
```

```python
from package import maths
maths.addition(2, 3)
```
