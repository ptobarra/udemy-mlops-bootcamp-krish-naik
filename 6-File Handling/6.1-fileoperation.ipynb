#### file operation - read and write files
file handling is a crucial part of any programming language. python provides built-in functions and methods to read and write to files, both text and binary. this lesson will cover the basics of file handlingf,. including reading and wirting text files and binary files.

```python
#### read whole file

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
## read a file line by line
with open('example.txt', 'r') as file:
    for line in file:
        print(line)
```

```python
## read a file line by line
with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())  # strip() removes the newline character
```

```python
## writing a file (overwriting)

with open('example.txt', 'w') as file:
    file.write('Hello World!\n')
    file.write('this is a new line.')
```

```python
## write a file (without overwriting)

with open('example.txt', 'a') as file:
    file.write('Append operation taking place!\n')
```

```python
## writing a list of lines to a file
lines = ['first line \n', 'second line \n', 'third line \n']

with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
## binary files

# writing to a binary file
data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
# reading a binary file
with open('example.bin', 'rb') as file:
    content = file.read()
    print(content)
```

```python
### read the content from a source text file and write it to a destination text file
# copying a text file
with open('example.txt', 'r') as source_file:
    content = source_file.read()

with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
# read a text file and count the number of lines, words, and characters
with open('example.txt', 'r') as source_file:
    content = source_file.read()

num_lines = len(content.splitlines())
num_words = len(content.split())    # split() with no args splits on any whitespace
num_chars = len(content)            # includes newline characters

print(content)
print(type(content))
print()
print(f"Lines: {num_lines}, Words: {num_words}, Characters: {num_chars}")
```

```python
# read a text file and count the number of lines, words, and characters
# counting lines, words, anb characters in a text file
def count_text_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = 'example.txt'
lines, words, characters = count_text_file(file_path)
print(f"Lines: {lines}, Words: {words}, Characters: {characters}")
```

the w+ mode in python is used to open as file for both reading and writing. if the file does not exist, it will be created. if the file exists, its content is truncated (i.e., the file is overwritten)

```python
### writing and then reading a file

with open('example.txt', 'w+') as file:
    file.write('Hello world\n')
    file.write('This is a new line\n')

    ## move the file cursor to the beginning
    # file.seek(0)

    ## read the content of the file
    content = file.read()
    print(content)
```

```python
with open('example.txt', 'w+') as file:
    file.write('Hello world\n')
    file.write('This is a new line\n')

    ## move the file cursor to the beginning
    file.seek(0)

    ## read the content of the file
    content = file.read()
    print(content)
```
